### Task 1: Calculate the eigenvalues of the matrices $A$,$B$ and assign them to *eig_A* and *eig_B*
**Hint:** You should calculate them by hand and insert them here. Alternatively, you may solve this in Julia, but without the use of any extra packages.

```julia
eig_A = 0
eig_B = 0

### BEGIN SOLUTION
# 1. Calculate the trace (sum of diagonals) and determinant
trace_A = A[1,1] + A[2,2]
det_A = (A[1,1] * A[2,2]) - (A[1,2] * A[2,1])

# 2. Use the quadratic formula for the characteristic equation: λ² - tr(A)λ + det(A) = 0
discriminant = trace_A^2 - 4 * det_A

# 3. Calculate both eigenvalues (roots of the polynomial)
lambda_1 = (trace_A + sqrt(discriminant)) / 2
lambda_2 = (trace_A - sqrt(discriminant)) / 2

println("Eigenvalues of Matrix A: ", lambda_1, ", ", lambda_2)
eig_A = [lambda_1, lambda_2]

# Similarly, for matrix
trace_B = B[1,1] + B[2,2]
det_B = (B[1,1] * B[2,2]) - (B[1,2] * B[2,1])

# 2. Use the quadratic formula for the characteristic equation: λ² - tr(B)λ + det(B) = 0
discriminant = trace_B^2 - 4 * det_B

# 3. Calculate both eigenvalues (roots of the polynomial)
lambda_1 = (trace_B + sqrt(Complex(discriminant))) / 2
lambda_2 = (trace_B - sqrt(Complex(discriminant))) / 2

println("Eigenvalues of Matrix B: ", lambda_1, ", ", lambda_2)
eig_B = [lambda_1, lambda_2]
### END SOLUTION
```

### Task 2: Calculate the matrix exponential function of matrix A and B (by hand) [ exp(A\*t) and exp(B\*t) ]

**Hint 1:** You should calculate them by hand and insert them here. Alternatively, you may solve this in Julia, but without the use of any extra packages.

**Hint 2:** The function should at the end return a 2x2 matrix 

**Important: Don't use exp() function**

```julia
#Here is an example how the functions should look like
function example(t)
    return [1*t 2; 3 4*t]
end
```

```julia
function matrixExponent(A, t)
    ### BEGIN SOLUTION
    
    # Initialize the result and the current term as the 2x2 Identity matrix
    result = [1.0 0.0; 0.0 1.0]
    term = [1.0 0.0; 0.0 1.0]
    
    # Run the Taylor Series expansion. 
    # Because A is nilpotent (A^2 = 0), this loop actually converges 
    # to the exact answer after just 1 iteration, but we can leave it 
    # at a higher number for consistency.
    for i in 1:50
        term = (term * A * t) / i
        result += term
    end
    
    return result
    ### END SOLUTION
end
```

```julia
function state_response(t, A, x0)
    ### BEGIN SOLUTION
    return matrixExponent(t, A) * x0
    ### END SOLUTION
end
```

### Task 4: Plot the resulting time series with plots.jl for a given time interval into a common plot

Note: Take care of proper formatting of the plot. In particular, add correct axis labeling and a legend to be able to differentiate the time series.

```julia
import Pkg
Pkg.add("Plots")
using Plots
```

```julia
# time points
t_values = 0:0.01:5
N = size(t_values)[1]

x0 = [1; 1]

# fill with time series for matrix A and B respectively
xt_A = zeros((N, 2));
xt_B = zeros((N, 2))
```

```julia
### BEGIN SOLUTION
for i in 1:N
    xt_A[i, :] = state_response(t_values[i], A, x0)
    xt_B[i, :] = state_response(t_values[i], B, x0)
end
### END SOLUTION
```

```julia
# plot for A
# Extract state variables for system A
x1_A = xt_A[:, 1]
x2_A = xt_A[:, 2]

# Plot system A
plot(t_values, x1_A, label="x₁(t)", lw=2)
plot!(t_values, x2_A, label="x₂(t)", lw=2)

xlabel!("Time")
ylabel!("State Variables")
title!("State Response of System A")
plot!(grid = true)
```

```julia
# plot for B
# Extract state variables for system B
x1_B = xt_B[:, 1]
x2_B = xt_B[:, 2]

# Plot system B
plot(t_values, x1_B, label="x₁(t)", lw=2)
plot!(t_values, x2_B, label="x₂(t)", lw=2)

xlabel!("Time")
ylabel!("State Variables")
title!("State Response of System B")
plot!(grid = true)
```
